n = int(input())

for i in range(1, n + 1):
    for j in range(1, n + 1):
        a = f'{i:03b}'
        b = f'{j:03b}'
        c = f'{i * j:06b}'
        print(f'{a}×{b}={c}', end='    ')
    print()

使用者輸入：請輸入正整數$n$，其中$1\le n\le 9$，輸出二進位$nn$乘法表。格式如如Tranclass的圖片Exercise20261008.png，以$n=5$為例。                                                                                                         
                                                                                                            
1. 只能使用while。                                                                                            
3. 字串格式化只能使用 f-string 對齊。   
n = int(input())

i = 1
while i <= n:
    j = 1
    while j <= n:
        a = f'{i:03b}'
        b = f'{j:03b}'
        c = f'{i * j:06b}'
        print(f'{a}×{b}={c}', end='    ')
        j += 1
    print()
    i += 1
